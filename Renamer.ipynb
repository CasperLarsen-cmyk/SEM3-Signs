```python
from os import listdir
import os

Folder = "./OURdata/Contrast/OurCutsCopyExWhite/"
files = listdir(Folder)

for i, file in enumerate(files):
    exStart = file.rfind('.')
    ext = file[exStart:]
    os.rename(Folder+file, f"{Folder}{i:0>3}ExWhite{ext}") 
```

```python
import cv2
import numpy as np
from os import listdir
import os

Folder = "./OURdata/OurCutsCopyExWhite/"
files = listdir(Folder)

B = +50

for file in files:
    filepath = os.path.join(Folder, file)

    image = cv2.imread(filepath)
    if image is None:
        continue

    image = image.astype(np.int16)

    image[image > 255 - B] = 255
    image[image < -B] = 0

    mask = (image > -B) & (image < 255 - B)
    image[mask] += B

    image = np.clip(image, 0, 255).astype(np.uint8)

    cv2.imwrite(filepath, image)
```
